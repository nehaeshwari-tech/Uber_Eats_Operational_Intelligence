```python
# Load raw datasets
customers = pd.read_csv(project_path / "data/raw/customers.csv")
orders = pd.read_csv(project_path / "data/raw/orders.csv")
payments = pd.read_csv(project_path / "data/raw/payments.csv")
deliveries = pd.read_csv(project_path / "data/raw/deliveries.csv")
restaurants = pd.read_csv(project_path / "data/raw/restaurants.csv")

# Inspect important columns and sample records
raw_datasets = {
    "Customers": customers,
    "Orders": orders,
    "Payments": payments,
    "Deliveries": deliveries,
    "Restaurants": restaurants
}

for name, df in raw_datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("Columns:", df.columns.tolist())
    display(df.head(3))

# Check tip amount details
print("\nTIP AMOUNT SUMMARY")
print(payments["tip_amount"].describe())

print("\nMissing tip values:", payments["tip_amount"].isnull().sum())
print("Zero tip values:", (payments["tip_amount"] == 0).sum())
```

```python
# Merge customer, order, payment and delivery information

integrated_data = orders.merge(
    customers,
    on="customer_id",
    how="left"
)

integrated_data = integrated_data.merge(
    payments,
    on="order_id",
    how="left"
)

integrated_data = integrated_data.merge(
    deliveries,
    on="order_id",
    how="left"
)

# Display integrated dataset
print("Integrated Dataset Shape:", integrated_data.shape)
print("\nColumns:", integrated_data.columns.tolist())

display(integrated_data.head())
```

```python
print("Integrated Dataset Information:")
integrated_data.info()

print("\nMissing Values:")
print(integrated_data.isnull().sum())

print("\nDuplicate Rows:")
print(integrated_data.duplicated().sum())
```

```python
# Integrate customer segmentation results

integrated_data = integrated_data.merge(
    customer_segments[["customer_id", "customer_cluster", "customer_dbscan_cluster"]],
    on="customer_id",
    how="left"
)

# Integrate restaurant segmentation results

integrated_data = integrated_data.merge(
    restaurant_segments[["restaurant_id", "restaurant_cluster", "restaurant_dbscan_cluster"]],
    on="restaurant_id",
    how="left"
)

# Validate segmentation integration

print("Updated Integrated Dataset Shape:", integrated_data.shape)

print("\nMissing Values in Cluster Columns:")
print(
    integrated_data[
        [
            "customer_cluster",
            "customer_dbscan_cluster",
            "restaurant_cluster",
            "restaurant_dbscan_cluster"
        ]
    ].isnull().sum()
)

display(integrated_data.head())
```

```python
# Select relevant sentiment columns
sentiment_data = reviews[
    ["order_id", "sentiment"]
].copy()

# Merge sentiment results with integrated dataset
integrated_data = integrated_data.merge(
    sentiment_data,
    on="order_id",
    how="left"
)

# Validate sentiment integration
print("Updated Integrated Dataset Shape:", integrated_data.shape)

print("\nMissing Sentiment Values:")
print(integrated_data["sentiment"].isnull().sum())

print("\nSentiment Distribution:")
print(integrated_data["sentiment"].value_counts())

display(integrated_data.head())
```

```python
# Convert order timestamp to datetime
integrated_data["order_timestamp"] = pd.to_datetime(
    integrated_data["order_timestamp"]
)

# Extract hour and date
integrated_data["order_hour"] = (
    integrated_data["order_timestamp"].dt.hour
)

integrated_data["order_date"] = (
    integrated_data["order_timestamp"].dt.date
)

# Analyze demand by city and hour
city_hourly_demand = (
    integrated_data
    .groupby(["city", "order_hour"])
    .size()
    .reset_index(name="total_orders")
)

# Display highest-demand city and hour combinations
city_hourly_demand = city_hourly_demand.sort_values(
    "total_orders",
    ascending=False
)

print("Top 10 City-Hour Demand Combinations:")
display(city_hourly_demand.head(10))
```

```python
# Analyze average tip amount by payment method
tip_by_payment = (
    integrated_data
    .groupby("payment_method")["tip_amount"]
    .agg(["mean", "count"])
    .reset_index()
)

tip_by_payment.columns = [
    "payment_method",
    "average_tip",
    "total_orders"
]

print("Average Tip by Payment Method:")
display(tip_by_payment)

# Analyze average tip amount by delivery conditions
tip_by_conditions = (
    integrated_data
    .groupby(["weather", "traffic"])["tip_amount"]
    .mean()
    .reset_index(name="average_tip")
)

print("\nAverage Tip by Weather and Traffic:")
display(tip_by_conditions)

# Analyze relationship between tip and numerical variables
print("\nCorrelation with Tip Amount:")
display(
    integrated_data[
        ["tip_amount", "order_amount", "distance_km",
         "delivery_time_minutes", "age"]
    ].corr()["tip_amount"].sort_values(ascending=False)
)
```
